```python
import random
import math

# Fungsi untuk menghitung nilai h (jumlah pasangan ratu yang saling menyerang)
def objective_function(board):
    h = 0
    for i in range(len(board)):
        for j in range(i + 1, len(board)):
            # Cek serangan pada baris yang sama (horizontal) dan diagonal
            if board[i] == board[j] or abs(board[i] - board[j]) == j - i:
                h += 1
    return h

# Fungsi untuk membuat state awal secara acak
def create_random_board(size=8):
    return [random.randint(0, size - 1) for _ in range(size)]

def hill_climbing():
    # Mulai dengan state awal yang acak
    current_board = create_random_board()
    current_h = objective_function(current_board)

    print(f"Initial board: {current_board} with h = {current_h}")

    while True:
        best_neighbor = current_board
        best_neighbor_h = current_h

        # Cari tetangga terbaik
        for col in range(len(current_board)):
            for row in range(len(current_board)):
                if current_board[col] == row:
                    continue  # Lewati posisi saat ini

                neighbor = list(current_board)
                neighbor[col] = row
                h = objective_function(neighbor)

                if h < best_neighbor_h:
                    best_neighbor_h = h
                    best_neighbor = neighbor

        # Jika tidak ada tetangga yang lebih baik, kita terjebak di local optimum
        if best_neighbor_h >= current_h:
            print(f"Stuck at local minimum. Final board: {current_board} with h = {current_h}")
            return current_board, current_h

        # Pindah ke tetangga terbaik
        current_board = best_neighbor
        current_h = best_neighbor_h
        print(f"Moved to board: {current_board} with h = {current_h}")

        if current_h == 0:
            print(f"Solution found! Board: {current_board} with h = {current_h}")
            return current_board, current_h

def simulated_annealing(temp=1000, cooling_rate=0.995):
    # Mulai dengan state awal yang acak
    current_board = create_random_board()
    current_h = objective_function(current_board)

    print(f"Initial board: {current_board} with h = {current_h}")

    t = 0
    while temp > 0.1:
        t += 1
        # Pilih tetangga secara acak
        next_board = list(current_board)
        col = random.randint(0, 7)
        row = random.randint(0, 7)
        next_board[col] = row

        next_h = objective_function(next_board)

        # Hitung perbedaan energi/nilai (delta E)
        delta_h = next_h - current_h

        # Jika langkah baru lebih baik, selalu terima
        if delta_h < 0:
            current_board = next_board
            current_h = next_h
        else:
            probability = math.exp(-delta_h / temp)
            if random.random() < probability:
                current_board = next_board
                current_h = next_h

        # Turunkan suhu
        temp *= cooling_rate

        if t % 100 == 0:
            print(f"Iteration {t}: temp={temp:.2f}, board={current_board}, h={current_h}")

        if current_h == 0:
            print(f"\nSolution found after {t} iterations! Board: {current_board} with h = {current_h}")
            return current_board, current_h

    print(f"\nSearch ended. Final board: {current_board} with h = {current_h}")
    return current_board, current_h


if __name__ == "__main__":
    print("--- Running Hill-Climbing Search ---")
    hill_climbing()

    print("\n\n--- Running Simulated Annealing Search ---")
    simulated_annealing()
```
